```python
# Question 44: Find sum of each nested list
print("\nQuestion 44: Find sum of each nested list")

dict_nested_list = {
    'matrix': [[1, 2], [3, 4]], 
    'vector': [5, 6, 7]
}

sums = {
    key: (sum(sum(item) if isinstance(item, list) else item for item in value))
    for key, value in dict_nested_list.items()
}
    
print(sums)
```
